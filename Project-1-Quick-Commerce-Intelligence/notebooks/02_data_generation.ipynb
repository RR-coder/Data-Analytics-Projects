```python
import pandas as pd
import numpy as np
from faker import Faker
from datetime import datetime, timedelta
import random

fake = Faker('en_IN')

print("Quick Commerce Data Generation Started")
```

```python
# Number of records
NUM_ORDERS = 100000

# Indian cities
cities = [
    "Delhi",
    "Mumbai",
    "Bangalore",
    "Hyderabad",
    "Pune",
    "Chennai",
    "Kolkata",
    "Ahmedabad"
]

# Product categories
categories = [
    "Dairy",
    "Snacks",
    "Beverages",
    "Fruits",
    "Personal Care",
    "Frozen Food",
    "Household"
]

# Payment methods
payment_methods = [
    "UPI",
    "Card",
    "Cash on Delivery",
    "Wallet"
]

# Order status
order_statuses = [
    "Delivered",
    "Cancelled"
]

print("Master data created successfully")
```

```python
# Generate realistic orders data

orders_data = []

for order_id in range(1, NUM_ORDERS + 1):

    city = random.choice(cities)

    # Peak hour simulation
    peak_hours = [8, 9, 10, 13, 14, 19, 20, 21]

    if random.random() < 0.7:
        hour = random.choice(peak_hours)
    else:
        hour = random.randint(0, 23)

    minute = random.randint(0, 59)

    order_time = datetime(
        2025,
        random.randint(1, 4),
        random.randint(1, 28),
        hour,
        minute
    )

    # Delivery duration
    delivery_minutes = max(10, int(np.random.normal(25, 8)))

    delivery_time = order_time + timedelta(minutes=delivery_minutes)

    # Cancellation probability
    if random.random() < 0.08:
        status = "Cancelled"
        delivery_time = None
    else:
        status = "Delivered"

    # Order values
    order_amount = round(random.uniform(100, 2500), 2)

    delivery_fee = round(random.uniform(10, 70), 2)

    discount = round(order_amount * random.uniform(0, 0.25), 2)

    final_amount = round(order_amount + delivery_fee - discount, 2)

    orders_data.append({
        "order_id": order_id,
        "city": city,
        "order_time": order_time,
        "delivery_time": delivery_time,
        "order_status": status,
        "payment_method": random.choice(payment_methods),
        "order_amount": order_amount,
        "delivery_fee": delivery_fee,
        "discount": discount,
        "final_amount": final_amount,
        "delivery_minutes": delivery_minutes
    })

print("Orders data generated successfully")
```

```python
orders_df = pd.DataFrame(orders_data)

orders_df.head()
```

```python
# Extract business features

orders_df["order_hour"] = orders_df["order_time"].dt.hour

orders_df["order_day"] = orders_df["order_time"].dt.day_name()

orders_df["order_month"] = orders_df["order_time"].dt.month_name()

# Weekend flag
orders_df["is_weekend"] = orders_df["order_time"].dt.weekday >= 5

# Peak hour flag
peak_hours = [8, 9, 10, 13, 14, 19, 20, 21]

orders_df["is_peak_hour"] = orders_df["order_hour"].isin(peak_hours)

# Delivery SLA breach
orders_df["sla_breach"] = orders_df["delivery_minutes"] > 30

print("Feature engineering completed")
```

```python
# Total orders
total_orders = len(orders_df)

# Total revenue
total_revenue = orders_df["final_amount"].sum()

# Average order value
avg_order_value = orders_df["final_amount"].mean()

# Cancellation rate
cancellation_rate = (
    orders_df["order_status"]
    .eq("Cancelled")
    .mean()
) * 100

print(f"Total Orders: {total_orders:,}")
print(f"Total Revenue: ₹{total_revenue:,.2f}")
print(f"Average Order Value: ₹{avg_order_value:.2f}")
print(f"Cancellation Rate: {cancellation_rate:.2f}%")
```

```python
city_performance = (
    orders_df
    .groupby("city")
    .agg({
        "order_id": "count",
        "final_amount": "sum",
        "delivery_minutes": "mean"
    })
    .reset_index()
)

city_performance.columns = [
    "city",
    "total_orders",
    "total_revenue",
    "avg_delivery_time"
]

city_performance.sort_values(
    by="total_revenue",
    ascending=False
)
```

```python
# Save generated dataset

orders_df.to_csv(
    "../data/synthetic/quick_commerce_orders.csv",
    index=False
)

print("Dataset exported successfully")
```
